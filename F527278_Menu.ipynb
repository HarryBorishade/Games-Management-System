```python
#f527278 13/01/2026

def get_all_games():

  all_games = []
  with open('Video_Game_Info.txt','r') as VideoGames, open('Board_Game_Info.txt','r') as BoardGames:
      next(VideoGames)
      next(BoardGames)
      for line in VideoGames:
          Vgames = line.strip().split(",")
          all_games.append(Vgames[1])
      for line in BoardGames:
          Bgames = line.strip().split(",")
          all_games.append(Bgames[1])


  return list(set(all_games))


def get_all_gameIDs():
    all_games = []
    with open('Video_Game_Info.txt','r') as VideoGames, open('Board_Game_Info.txt','r') as BoardGames:
        next(VideoGames)
        next(BoardGames)
        for line in VideoGames:
            Vgames = line.strip().split(",")
            all_games.append(Vgames[0])
        for line in BoardGames:
            Bgames = line.strip().split(",")
            all_games.append(Bgames[0])


    return list(set(all_games))


def get_all_genres():
    all_genres = []
    with open('Video_Game_Info.txt','r') as VideoGames, open('Board_Game_Info.txt','r') as BoardGames:
        next(VideoGames)
        next(BoardGames)
        for line in VideoGames:
            Vgames = line.strip().split(",")
            all_genres.append(Vgames[3])
        for line in BoardGames:
            Bgames = line.strip().split(",")
            all_genres.append(Bgames[3])

    return set(all_genres)

```

```python
#F527278 10/12/2025


from IPython.core.display import Video
import ipywidgets as widgets
from ipywidgets import Button
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import time

def Display_Search_Video_Game(search):
    gameList = []
    with open('Video_Game_Info.txt','r') as files:

        for line in files:
            games = [g.strip() for g in line.strip().split(",")]
            #strips all items into a value of the list

            if games[1] == "Name" or games[3] =="Genre":
                continue

            if games[1].lower() == search or games[3].lower() == search:
                gameList.append(games)
            else:
                pass

    if gameList == []:
        return "Void"
        # checks to see if the game searched is present in the file
    else:
        return gameList




def Display_Search_Board_Game(search):
    gameList = []
    with open('Board_Game_Info.txt','r') as files:

      for line in files:
        games = [g.strip() for g in line.strip().split(",")]
        #strips all items into a value of the list

        if games[1] == "Name" or games[3] =="Genre":
          continue

        if games[1].lower() == search or games[3].lower() == search:
          gameList.append(games)
        else:
          pass

    if gameList == []:
      return "Void"
      # checks to see if the game searched is present in the file
    else:
      return gameList





def on_videoGameButton_clicked(b):
    typedtext = search_box.value.lower()

    if typedtext == "":
        clear_output(wait=True)
        display(GameMenuWithHeader, search_box, VideoSearchButton, BoardSearchButton,AllGamesButton)
        display(Please_search_box)
        return


    Games = (Display_Search_Video_Game(typedtext))
    clear_output(wait=True)
    display(GameMenuWithHeader, search_box, VideoSearchButton, BoardSearchButton,AllGamesButton)

    # passes value of search box to function
    if Games == "Void":
        display(Game_not_found_box)
        return
    else:
        Availability = check_game_availability(Games)
        #Availability[0] is the list of available Video Games
        #Availability[1] is the list of unavailable Video Games
        VideoGamesTable(Availability[0],Availability[1])


def on_boardGameButton_clicked(b):
    typedtext = search_box.value.lower()

    if typedtext == "":
        clear_output(wait=True)
        display(GameMenuWithHeader, search_box, VideoSearchButton, BoardSearchButton,AllGamesButton)
        display(Please_search_box)
        return

    Games = (Display_Search_Board_Game(typedtext))
    clear_output(wait=True)
    display(GameMenuWithHeader, search_box, VideoSearchButton, BoardSearchButton,AllGamesButton)

    # passes value of search box to function
    if Games == "Void":
        display(Game_not_found_box)
        return
    else:
        Availability = check_game_availability(Games)
        #Availability[0] is the list of available Board Games
        #Availability[1] is the list of unavailable Board Games
        BoardGamesTable(Availability[0],Availability[1])


def check_game_availability(allgames):
    # takes games
    with open('Rental.txt', "r") as f:
        lines = f.readlines()[1:]
        # skip header

    rental_status = {}

    for line in lines:
        parts = line.strip().split(",")
        if len(parts) != 4:
            continue

        game_id, rental_date, return_date, customer_id = parts


        # Make sure game_id has a default entry
        if game_id not in rental_status:
            rental_status[game_id] = False   # assume available

        # If any record has no return date then game is rented
        if return_date.strip() == "":
            rental_status[game_id] = True

    available_games = []
    unavailable_games = []

    for record in allgames:
        game_id = str(record[0]).strip()
        is_rented = rental_status.get(game_id, False)

        if is_rented:
            unavailable_games.append(record)
        else:
            available_games.append(record)

    return available_games, unavailable_games



def VideoGamesTable(available,unavailable):

    #Display video game records in a matplotlib table.

    if not available and not unavailable:
        return "Not an available game"

    with open('Video_Game_Info.txt', 'r') as file:
        headerLine = file.readline()
        headerList = [h.strip() for h in headerLine.strip().split(",")]

        headerList.append('Availability')
        # To make a new field to show availiability


    # Build full table data: header + all records
    rows = [headerList]
    rows += [record + ["Available"] for record in available]
    rows += [record + ["Unavailable"] for record in unavailable]

    fig, ax = plt.subplots()
    ax.axis('tight')
    ax.axis('off')
    table = ax.table(cellText=rows, loc='center', cellLoc='center')
    table.scale(2.2, 4)
    plt.show()



def allGamesSearch():
    allgames = []

    with open('Video_Game_Info.txt','r') as VideoGames, open('Board_Game_Info.txt','r') as BoardGames:
        next(VideoGames)
        next(BoardGames)
        for line in VideoGames:
            games = [g.strip() for g in line.strip().split(",")]
            allgames.append(games)
        for line in BoardGames:
            games = [g.strip() for g in line.strip().split(",")]
            allgames.append(games)

    return allgames






def BoardGamesTable(available,unavailable):

    #Display board game records in a matplotlib table.
    #records should be a list of rows (each row is a list of fields).

    if not available and not unavailable:
        return "Not an available game"

    # Read header from file
    with open('Board_Game_Info.txt', 'r') as file:
        headerLine = file.readline()
        headerList = [h.strip() for h in headerLine.strip().split(",")]

        headerList.append('Availability')
        # To make a new field to show availiability


    # Build full table data: header + all records
    rows = [headerList]
    rows += [record + ["Available"] for record in available]
    rows += [record + ["Unavailable"] for record in unavailable]

    # Plot table
    fig, ax = plt.subplots()
    ax.axis('tight')
    ax.axis('off')
    table = ax.table(cellText=rows, loc='center', cellLoc='center')
    table.scale(2.2, 4)
    plt.show()


def on_allGamesButton_clicked(b):
    #checks the availability of all the games so it can be passed into the table fn
    clear_output(wait=True)
    display(GameMenuWithHeader, search_box, VideoSearchButton, BoardSearchButton, AllGamesButton)

    allGames = allGamesSearch()
    Availability = check_game_availability(allGames)

    VideoGamesTable(Availability[0], Availability[1])


Game_search_box_options = sorted(set(get_all_games()) | get_all_genres())
#Search box creation
search_box = widgets.Combobox(
    options= Game_search_box_options,
    value = None,
    description='Game :',
    disabled=False
)


#VideoGame Search button
VideoSearchButton = widgets.Button(
    description='Video Games',
    button_style='success'
)

#BoardGame Search button
BoardSearchButton = widgets.Button(
    description='Board Games',
    button_style='success'
)

# Button to output all games
AllGamesButton = widgets.Button(
    description='Show all games',
    button_style='info'
)


#General Message box

General_message_box = widgets.Label(
    value="",
    layout=widgets.Layout(
        border="4px solid green",
        border_radius="6px",
        padding="6px",
        width="250px"
    )
)
General_error_box = widgets.Label(
    value="",
    layout=widgets.Layout(
        border="4px solid red",
        border_radius="6px",
        padding="6px",
        width="150px"
    )
)





AllGamesButton.on_click(on_allGamesButton_clicked)
BoardSearchButton.on_click(on_boardGameButton_clicked)
VideoSearchButton.on_click(on_videoGameButton_clicked)

```

###Game Return

```python
#F527278 24/12/2025
from os import wait
from IPython.core.debugger import re
import subscriptionManager as smSL
import feedbackManager as fmSL
from datetime import date




def return_game(CustomerID,GameReturnID):
    CustomerID = CustomerID.lower()
    GameReturnID = GameReturnID.lower()


    if GameReturnID not in get_all_gameIDs():
        return "Invalid gameID"
        # checks if the game ID is valid

    updated = False
    newline = []

    with open('Rental.txt', "r") as file:
        header = next(file)
        # skip header
        for line in file:
            records = [g.strip() for g in line.strip().split(",")]
            if len(records) != 4:
                newline.append(line)
                continue

            game_id, rental_date, return_date, customer_id = records

            if game_id == GameReturnID and return_date == "":
                if customer_id != CustomerID:
                     return "Game has already returned"


                return_date = date.today().isoformat()
                newline.append(f"{game_id},{rental_date},{return_date},{customer_id}\n")
                updated = True
            else:
                newline.append(line)

    if updated == False:
        return "Game not found"
    else:
        file = open('Rental.txt', "w")
        file.write(header)
        file.writelines(newline)
        file.close()

    return "Successful Return"


def on_game_return_clicked(b):
    clear_output(wait=True)
    display(GameMenuWithHeader, User_return_search,Game_return_search, Game_return_button)

    typedCustomer_id = User_return_search.value.strip().lower()
    typedGame_id = Game_return_search.value.strip().lower()

    if not typedCustomer_id:
        display(Please_enter_customerID_box)
        return

    elif not typedGame_id:
        display(Please_enter_gameID_box)
        return


    else:
        result = return_game(typedCustomer_id, typedGame_id)

        if result != "Successful Return":
            print(result)
            return

    clear_output(wait=True)
    display(GameMenuWithHeader, Game_comment, Game_rating, Feedback_return)
    return





# Game reveiw function
def Game_review(game_id, rating, comment):
    Game_comment.layout.display = "flex"
    Game_rating.layout.display = "flex"
    Feedback_return.layout.display = "flex"
    fmSL.add_feedback(game_id, rating, comment, "Game_Feedback.txt")
    return

output = widgets.Output()


def on_feedback_return_clicked(b):
    clear_output(wait=True)
    display(GameMenuWithHeader, Game_comment, Game_rating, Feedback_return)

    rating_input = Game_rating.value.strip()
    comment_input = Game_comment.value.strip()
    game_id = Game_return_search.value.strip()

    # rating validation
    try:
        rating = int(rating_input)
        if rating < 1 or rating > 5:
            raise ValueError

    except ValueError:
        display(Please_Enter_Feedback_Rating)
        return


    # comment validation
    try:
        if not comment_input:
            raise ValueError
        comment = comment_input

    except ValueError:
        display(Please_Enter_feedback_Comment)
        return


    clear_output(wait=True)
    display(GameMenuWithHeader, User_return_search,Game_return_search, Game_return_button)

    Game_comment.layout.display = "None"
    Game_rating.layout.display = "None"
    Feedback_return.layout.display = "None"

    Game_review(game_id, rating, comment)


Game_return_button.on_click(on_game_return_clicked)
Feedback_return.on_click(on_feedback_return_clicked)
```

```python
#F527278 03/01/2026
from IPython.core.display import Video
import ipywidgets as widgets
from ipywidgets import Button
from IPython.display import display, clear_output
import matplotlib.pyplot as plt
import time
from datetime import date, datetime, timedelta
from collections import Counter


Ratings_inputs_box = widgets.VBox()
Ratings_input_rows = []

Average_rating_button = widgets.Button(
    description="Average Ratings",
    button_style="success"
)


Ratings_add_button = widgets.Button(
    description="Add Input",
    button_style="success"
)

Ratings_submit_button = widgets.Button(
    description="Submit All",
    button_style="primary"
)

Ratings_label = widgets.Label(
    value="Enter game names to get their average ratings:",
    layout = widgets.Layout(
        width="300px",
        padding = "8px",
        border = "1px solid lightgray",
        border_radius = "8px",
        margin = "5px 0px"
))


def create_input_row():
    text = widgets.Text(
        placeholder="Enter game name...",
    )

    delete_btn = widgets.Button(
        description="X",
        button_style="danger",
        layout=widgets.Layout(width="40px")
    )

    row = widgets.HBox([text, delete_btn])

    delete_btn.row = row
    delete_btn.on_click(delete_row_clicked)

    return row


def add_input_clicked(b):
    row = create_input_row()
    Ratings_input_rows.append(row)
    Ratings_inputs_box.children = Ratings_input_rows



def submit_clicked(b):
    clear_output(wait=True)
    display(GameMenuWithHeader, Ratings_label, Ratings_add_button, Ratings_inputs_box, Ratings_submit_button)
    values = []

    for row in Ratings_input_rows:
        game_name = row.children[0].value.strip().lower()

        if game_name == "":
            continue

        if game_name not in get_all_games():
            continue

        values.append(game_name)

    if len(values) == 0:
        print("No valid game names entered.")
    else:
        Check_rating_average(values)

def delete_row_clicked(b):
    row = b.row
    if row in Ratings_input_rows:
        Ratings_input_rows.remove(row)
        Ratings_inputs_box.children = Ratings_input_rows

def average_rating_clicked(b):
    clear_output(wait=True)
    display(GameMenuWithHeader, Ratings_label, Ratings_add_button, Ratings_inputs_box, Ratings_submit_button)
    return


def Check_rating_average(GameNames):
    GameNames = [g.lower().strip() for g in GameNames]
    game_to_ids = {name: [] for name in GameNames}

    # Match game names to IDs
    with open('Board_Game_Info.txt', 'r') as Board:
        next(Board)
        for line in Board:
            records = [g.strip() for g in line.split(",")]
            if len(records) != 5:
                continue

            game_id = records[0]
            game_name = records[1].lower()

            if game_name in game_to_ids:
                game_to_ids[game_name].append(game_id)

    with open('Video_Game_Info.txt', 'r') as Video:
        next(Video)
        for line in Video:
            records = [g.strip() for g in line.split(",")]
            if len(records) != 5:
                continue

            game_id = records[0]
            game_name = records[1].lower()

            if game_name in game_to_ids:
                game_to_ids[game_name].append(game_id)

    feedback_list = fmSL.load_feedback()

    games_with_feedback = []
    average_ratings = []
    games_without_feedback = []

    # Explicit rating collection and appending
    for game_name, ids in game_to_ids.items():

        ratings = []

        for feedback in feedback_list:
            if feedback['GameID'] in ids:
                ratings.append(feedback['Rating'])

        if len(ratings) == 0:
            games_without_feedback.append(game_name)

        else:
            total_rating = sum(ratings)
            rating_count = len(ratings)
            average = total_rating / rating_count

            games_with_feedback.append(game_name)
            average_ratings.append(average)

    if len(games_without_feedback) > 0:
        print(
            "No feedback available for:",
            ", ".join(games_without_feedback)
        )

    if len(games_with_feedback) > 0:
        Average_rating_graphing(games_with_feedback, average_ratings)
    else:
        print("No games with feedback to display.")


def Average_rating_graphing(game_names, average_ratings):
    bar_count = len(game_names)
    width = max(6, bar_count * 1.2)

    plt.figure(figsize=(width, 8))
    plt.bar(game_names, average_ratings)

    for i in range(len(average_ratings)):
        plt.text(i, average_ratings[i] + 0.05,
                 f"{average_ratings[i]:.2f}",
                 ha='center')

    plt.title("Average Ratings per Game")
    plt.xlabel("Game Name")
    plt.ylabel("Average Rating")
    plt.ylim(0, 7)
    plt.show()


2# Filter Feedback

Filter_feedback_button = widgets.Button(
    description="Filter Feedback",
    button_style="success"
)

def on_Filter_feedback_button_clicked(b):
    clear_output(wait=True)
    display(GameMenuWithHeader, comment_feedback_text, comment_feedback_button)

def feedback_filter(Comment_text):
    plt.close('all')
    clear_output(wait=True)
    display(GameMenuWithHeader, comment_feedback_text, comment_feedback_button)
    feedback_list = fmSL.load_feedback()

    if len(feedback_list) == 0:
        return 'No feedback'

    Comment_text = Comment_text.strip().lower()
    games = []

    for feedback in feedback_list:
        existing_comment = feedback['Comments'].strip().lower()

        if Comment_text in existing_comment:
            games.append(feedback)

    return games

def draw_table(games):

    if not games:
        print('No feedback fits your criteria')
        return

    fig, ax = plt.subplots(figsize=(9, 3))
    ax.axis("off")
    columns = ["GameID", "Rating", "Comments"]
    table_data = [[game["GameID"], game["Rating"],
                 game["Comments"]] for game in games]

    table = ax.table(
      cellText=table_data,
      colLabels=columns,
      cellLoc="center",
      loc="center"
  )

    table.auto_set_font_size(False)
    table.set_fontsize(11)
    table.scale(1, 1.6)

    for col in range(len(columns)):
        cell = table[(0, col)]
        cell.set_facecolor("#2C3E50")
        cell.get_text().set_color("white")
        cell.get_text().set_weight("bold")

    for row in range(1, len(table_data) + 1):
        for col in range(len(columns)):
            cell = table[(row, col)]

    plt.show()


comment_feedback_text = widgets.Text(
  placeholder="Type to filter comments...",
  description="Search:",
  layout=widgets.Layout(width="300px")
)


def on_comment_feedback_button_clicked(b):
    search_text = comment_feedback_text.value
    games = feedback_filter(search_text)
    draw_table(games)



comment_feedback_button = widgets.Button(
  description="Submit",
  button_style="success"
)

comment_feedback_button.on_click(on_comment_feedback_button_clicked)
Filter_feedback_button.on_click(on_Filter_feedback_button_clicked)



3# Filter rentals top 10% by time period


Filter_rental_button = widgets.Button(
    description="Filter Rentals",
    button_style="success"
)

def on_Filter_rental_button_clicked(b):
    clear_output(wait=True)
    display(GameMenuWithHeader, period_dropdown)
    update_rental_frequency()


def filter_rentals_by_time_period(rentals, days):
    cutoff_date = datetime.now() - timedelta(days=days)

    return [
        rental for rental in rentals
        if rental["RentalDate"] >= cutoff_date
    ]

def get_rental():
    rentals = []

    with open('Rental.txt','r') as file:
        next(file)
        for line in file:
            records = [g.strip() for g in line.strip().split(",")]
            if len(records) != 4:
                continue

            game_id, rental_date, return_date, customer_id = records
            rentals.append({
          "GameID": game_id,
          "RentalDate": datetime.strptime(rental_date, "%Y-%m-%d"),
          "ReturnDate": datetime.strptime(return_date, "%Y-%m-%d") if return_date else None,
          "CustomerID": customer_id
      })

    return rentals


def calculate_rental_frequency(rentals):
    game_ids = [rental["GameID"] for rental in rentals]
    return Counter(game_ids)


def identify_top_10_percent_games(frequency):
    if not frequency:
        return set()

    sorted_counts = sorted(frequency.values(), reverse=True)
    cutoff_index = max(1, int(len(sorted_counts) * 0.1))
    cutoff_value = sorted_counts[cutoff_index - 1]

    return {
    game_id for game_id, count in frequency.items()
    if count >= cutoff_value
    }



def draw_rental_frequency_chart(frequency, top_games):
    clear_output(wait=True)
    display(GameMenuWithHeader, period_dropdown)

    if not frequency:
        print("No rental data for selected period.")
        return

    game_ids = list(frequency.keys())
    counts = list(frequency.values())

    fig, ax = plt.subplots(figsize=(10, 4))
    bars = ax.bar(game_ids, counts)

    # Highlight top 10%
    for bar, game_id in zip(bars, game_ids):
        if game_id in top_games:
            bar.set_edgecolor("black")
            bar.set_linewidth(3)

    ax.set_title("Rental Frequency by Game ID")
    ax.set_xlabel("Game ID")
    ax.set_ylabel("Number of Rentals")
    ax.tick_params(axis="x", rotation=45)

    plt.tight_layout()
    plt.show()


def update_rental_frequency(change=None):
    rentals = get_rental()

    filtered_rentals = filter_rentals_by_time_period(
        rentals,
        period_dropdown.value
    )

    frequency = calculate_rental_frequency(filtered_rentals)
    top_games = identify_top_10_percent_games(frequency)

    draw_rental_frequency_chart(frequency, top_games)
    scores = calculate_pruning_scores(filtered_rentals)
    pruned_games = identify_prune_candidates(scores)
    display_pruning_summary(pruned_games)


def get_average_ratings():
    feedback_list = fmSL.load_feedback()
    ratings_by_game = {}

    for feedback in feedback_list:
        game_id = feedback["GameID"]
        ratings_by_game.setdefault(game_id, []).append(feedback["Rating"])

    return {
        game_id: sum(ratings) / len(ratings)
        for game_id, ratings in ratings_by_game.items()
    }

def calculate_pruning_scores(rentals, rental_weight=0.6,
                             rating_weight=0.4):

    rental_frequency = calculate_rental_frequency(rentals)
    average_ratings = get_average_ratings()
    scores = {}

    for game_id in rental_frequency:
        scores[game_id] = (
            rental_weight * rental_frequency.get(game_id, 0)
            + rating_weight * average_ratings.get(game_id, 0)
        )

    return scores

def identify_prune_candidates(scores, prune_percent=0.2):
    if not scores:
        return []

    sorted_games = sorted(scores.items(), key=lambda x: x[1])
    cutoff = int(len(sorted_games) * prune_percent)

    return sorted_games[:cutoff]


def display_pruning_summary(pruned_games):
    if not pruned_games:
        print("No games qualify for pruning.")
        return

    print("INVENTORY PRUNING RECOMMENDATIONS\n")

    for game_id, score in pruned_games:
        print(
            f"Game ID: {game_id} | "
            f"Pruning Score: {score:.2f} | "
            "Reason: Low rental demand and/or low ratings"
        )



period_dropdown = widgets.Dropdown(
    options={
        "Last 30 Days": 30,
        "Last 60 Days": 60,
        "Last 1 Year": 365
    },
    description="Period:"
)

period_dropdown.observe(update_rental_frequency, names="value")

Filter_rental_button.on_click(on_Filter_rental_button_clicked)



Average_rating_button.on_click(average_rating_clicked)
Ratings_add_button.on_click(add_input_clicked)
Ratings_submit_button.on_click(submit_clicked)
Filter_rental_button.on_click(on_Filter_rental_button_clicked)

```

###<u><b>Briefly discuss five security code issues.</b></u>

#### <b>Input validation:</b>

* Inputs need to be validated based on what they will be used for e.g. Making sure the right type of data is passed into the variable because this could result in mathematical operations being passed on words which would result in an error.  

<u>Solution:</u>
* Whitelist validations by defining acceptable input patters and rejecting everything else.


#### <b> Hard Coding sensitive data in a program:</b>

* This poses a security code issue since if the code is shared to another person, the person accessing the code will have access to the sensitive data which can be used for malicious purposes.

<u>Solution:</u>

* Storing the sensitive data in a sepearate document will prevent them from being accessed.

#### <b> Predictable Randomness</b>
* If a recognised random seed is used to generate numbers or other things. It could result in securely encrypted operations being guessed.

<u>Solution</u>
* Use a secret python module, and avoid fixed seed in value productions.

#### <b> Global Variables </b>
* The variables may be altered by accident inside of a functions and this may disrupt the rest of the code causing errors. In addition to this it can make sensitive data made public and allow people to access it.

<u>Solution</u>
* Minimise the use of global variables and parameters passed.


#### <b> Loops without clearly definable and or satisfiable criterias </b>
* Loops without clear, satisfiable exit conditions can cause infinite loops. This can reduce system availabilty and make the system less robost and prone to crashes.

<u> Solutions:</u>
* Every loops needs a definable way to terminate.
